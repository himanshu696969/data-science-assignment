# Movie Ratings Analysis
A small sample notebook for practicing Pandas and visualization.

```python
import pandas as pd
ratings = pd.DataFrame({
    'movie': ['Movie A', 'Movie B', 'Movie C'],
    'rating': [4.5, 3.8, 4.2]
})
ratings
```

```python
import matplotlib.pyplot as plt
ratings.plot(kind='bar', x='movie', y='rating', legend=False)
plt.title('Movie Ratings Comparison')
plt.xlabel('Movie')
plt.ylabel('Rating')
plt.tight_layout()
plt.show()
```
